```python
import pandas as pd

df = pd.read_csv(r'..\data\W158_select.csv')

df.info()
```

```python
dep_var_list = [
    'CONF_a_W158',
    'CONF_b_W158',
    'CONF_c_W158',
    'CONF_d_W158',
    'CONF_e_W158',
    'CONF_i_W158',
    'CONF_j_W158'
    ]

print(dep_var_list)
```

```python
df[dep_var_list].duplicated().sum()
```

```python
patterns = df[dep_var_list].value_counts().reset_index(name="count")
patterns["pct"] = (patterns["count"] / len(df[dep_var_list]) * 100).round(2)
patterns["cum_pct"] = patterns["pct"].cumsum().round(2)

print("Unique patterns:", len(patterns))
print(patterns.head(20).to_string(index=False))
```
