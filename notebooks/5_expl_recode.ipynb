```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

group_var = "F_PARTYSUM_FINAL"  

# 1. Include the grouping variable in the subset and rename dependent variables
subset_df = df[[group_var] + dep_var_list].rename(columns=dep_labels)

# 2. Reshape from wide to long format for Seaborn's hue mapping
long_df = pd.melt(
    subset_df,
    id_vars=[group_var],
    var_name="Variable",
    value_name="Score",
)

# 3. Determine column ordering based on overall means
sorted_columns = (
    long_df.groupby("Variable")["Score"]
    .mean()
    .sort_values(ascending=False)
    .index
)

# 4. Plot with hue mapping for grouped bars
fig, ax = plt.subplots()
sns.barplot(
    data=long_df,
    x="Variable",
    y="Score",
    hue=group_var,
    order=sorted_columns,
    errorbar=("ci", 95),
    capsize=0.1,
    ax=ax,
)

# Annotate each bar group with mean values
for container in ax.containers:
    ax.bar_label(container, fmt="%.2f", padding=3)

ax.set_ylim(1, 4)
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```
