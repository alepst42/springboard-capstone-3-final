In documentation, value labels are:

1	A great deal of confidence
2	A fair amount of confidence
3	Not too much confidence
4	No confidence at all
99	Don't know/Refused/Web blank

Therefore, will reverse code

```python
import pandas as pd

df = pd.read_csv(r'..\data\W158_select_nan.csv')

df.info()
```

```python
dep_var_list = [
    'CONF_a_W158',
    'CONF_b_W158',
    'CONF_c_W158',
    'CONF_d_W158',
    'CONF_e_W158',
    'CONF_i_W158',
    'CONF_j_W158'
    ]

print(dep_var_list)
```

```python
#Rename variables

df = df.join(df[dep_var_list].add_suffix("_r"))

df.info()
```

```python
dep_var_list_r = [
    'CONF_a_W158_r',
    'CONF_b_W158_r',
    'CONF_c_W158_r',
    'CONF_d_W158_r',
    'CONF_e_W158_r',
    'CONF_i_W158_r',
    'CONF_j_W158_r'
    ]

print(dep_var_list_r)
```

```python
# Recode

mapping = {1.0: 4.0, 2.0: 3.0, 3.0: 2.0, 4.0: 1.0}
df[dep_var_list_r] = df[dep_var_list_r].replace(mapping)
```

```python
df_r = df.drop(columns=dep_var_list)
df_r.info()
```
