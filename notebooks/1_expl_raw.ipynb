```python
import pandas as pd
```

```python
df = pd.read_csv(r'..\data\W158_Oct24\W158_Oct24\ATP W158.csv')
df.info()
```

```python
string_cols = df.select_dtypes(include=["object", "string"]).columns.tolist()
print(string_cols)
```

```python
total_missing = df.isnull().sum().sum()
print(total_missing)
```

```python
# Distribution of missing values per row
missing_per_row = df.isnull().sum(axis=1)

# Summary statistics of row-level missingness
print(missing_per_row.describe())
```

```python
import matplotlib.pyplot as plt

plt.hist(missing_per_row)
```

```python
print(missing_per_row.value_counts())
```

```python
missing_per_col = df.isnull().sum()
print(missing_per_col.describe())
```

```python
plt.hist(missing_per_col, bins = 25)
```

```python
print(missing_per_col.value_counts())
```

```python
unique_counts = df.nunique(dropna=False)
print(unique_counts)
```

```python
unique_counts.describe()
```

```python
unique_counts.value_counts()
```
